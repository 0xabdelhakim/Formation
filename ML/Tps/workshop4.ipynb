## Dataset Context
The California Housing dataset contains data from the 1990 California census, with about 20,640 block groups (small areas). 
The target variable is the median house value (in hundreds of thousands of dollars), which we might predict in regression tasks.
The features include:
- MedInc: Median income (float)
- HouseAge: Median house age (float)
- AveRooms: Average rooms per household (float)
- AveBedrms: Average bedrooms per household (float)
- Population: Block group population (float)
- AveOccup: Average occupants per household (float)
- Latitude: Block group latitude (float)
- Longitude: Block group longitude (float)
This data represents real estate patterns in California and is used to explore housing prices based on socioeconomic and geographic factors. Think of it as realtor analysis: inspecting features to see what drives home values, like income or location.

### Step 1: Loading the Dataset - Using Scikit-Learn Again
We load from sklearn.datasets, similar to Iris, but now a regression dataset. This reinforces loading while introducing larger numerical data, perfect for NumPy.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_california_housing
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
housing = fetch_california_housing()

# Create DataFrame
housing_df = pd.DataFrame(housing.data, columns=housing.feature_names)
housing_df['MedHouseVal'] = housing.target  # Add target

# Display first 5 rows
housing_df.head()
```

# Output interpretation
Table with columns like MedInc ~8.32, HouseAge ~41, etc., MedHouseVal ~4.52 (452k). Notice all numerical, no categories yet – ideal for NumPy and correlations.

### Step 2: Introducing NumPy Arrays - Efficient Numerical Storage
NumPy arrays are faster for math than lists. We teach now because housing data is all numbers, and ML often uses arrays for inputs. Convert DF to array contextually.

```python
# Convert DataFrame to NumPy array
housing_array = housing_df.to_numpy()

# Print type and shape
print(type(housing_array))
print(housing_array.shape)

# First row
print(housing_array[0])
```

# Output interpretation
<class 'numpy.ndarray'>, (20640, 9) – 20k+ rows, 9 cols. First row: array of floats like [8.3252, 41., ... , 4.526]. Notice it's 2D: rows as samples, cols as features + target.

### Step 3: Basic NumPy Operations - Arithmetic and Aggregation
Perform math on whole arrays. This shows why NumPy: vectorized ops faster than loops (recall last week's loop warning).

```python
# Select MedInc column (index 0)
med_inc = housing_array[:, 0]

# Mean and std
mean_inc = np.mean(med_inc)
std_inc = np.std(med_inc)
print('Mean Median Income:', mean_inc)
print('Std Dev:', std_inc)

# Normalize (simple scaling)
normalized_inc = (med_inc - mean_inc) / std_inc
print(normalized_inc[:5])
```

# Output interpretation
Mean ~3.87, Std ~1.90. Normalized: first ~2.34 (above avg), others vary. This demos array-wide math without loops – efficient for large data.

### Step 4: Slicing NumPy Arrays - Accessing Subsets
Slicing like lists but multi-dimensional. Needed for selecting features or samples.

```python
# First 5 rows, first 3 columns
subset = housing_array[:5, :3]
print(subset)

# All rows, Latitude and Longitude (indices 6,7)
locations = housing_array[:, 6:8]
print(locations.shape)
```

# Output interpretation
5x3 array of numbers. (20640, 2) for locations. Notice : for all, comma for dims – like matrix slicing.

### Step 5: Advanced Inspection - Correlation Analysis
Use .corr() to see feature relationships. After NumPy because correlations are numerical, could use np.corrcoef but Pandas easier.

```python
# Compute correlation matrix
corr_matrix = housing_df.corr()

# Display
print(corr_matrix)
```

# Output interpretation
Matrix: MedInc vs MedHouseVal ~0.68 (strong positive), Latitude vs Longitude ~-0.92 (strong negative, as expected geographically). Values -1 to 1: high abs means related.

### Step 6: Visualizing Correlations - Heatmap
Heatmap for easy spotting of strong correlations.

```python
# Plot heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

# Output interpretation
Color grid: Red positive, blue negative. Annot shows values. Notice MedInc red with target – key predictor. AveRooms/AveBedrms correlated – maybe redundant.

### Step 7: Detecting Outliers - Using Box Plots and Scatter
Visuals to spot anomalies. Builds on boxplots from last week.

```python
# Box plot for MedHouseVal
plt.figure(figsize=(8, 5))
sns.boxplot(x=housing_df['MedHouseVal'])
plt.title('House Value Distribution')
plt.show()

# Scatter for MedInc vs MedHouseVal
plt.figure(figsize=(8, 5))
sns.scatterplot(x='MedInc', y='MedHouseVal', data=housing_df)
plt.title('Income vs House Value')
plt.show()
```

# Output interpretation
Box: Median ~1.8, outliers above ~5 (capped?). Scatter: Positive trend, but ceiling at 5 – data artifact. Spots clusters or extremes.

### Step 8: Functions for Inspection - Reusable Code
Wrap correlation and plot in function.

```python
# Define function
def plot_correlations(df, target):
    corr = df.corr()
    plt.figure(figsize=(10, 8))
    sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
    plt.title(f'Correlations including {target}')
    plt.show()

# Call
plot_correlations(housing_df, 'MedHouseVal')
```

# Output interpretation
Same heatmap. Function allows reuse on subsets or other DFs – professional.

## ⚠️ Common Mistakes Box
- Lists vs arrays: Math on lists needs loops; arrays vectorize.
- Wrong slicing: housing_array[:,0] for column, not [0,:].
- Ignoring correlations: Miss multicollinearity (high corr features).
- Plotting without figsize: Tiny unreadable plots.
- Outliers unhandled: Can skew means, models – always visualize.
- Function no params: Hardcoded, not reusable.
