## Dataset Context

The Digits dataset consists of 1,797 grayscale images of handwritten digits (0 through 9), each 8x8 pixels.
The target variable is the digit class (0-9), which we might classify in ML tasks.
The features are the 64 pixel values (0-16 intensity) per image, represented as a flattened array.
This data simulates real optical recognition problems, like reading zip codes. Think of it as computer vision basics: inspecting pixel patterns to see how digits differ, preparing for tasks like dimensionality reduction later.


### Step 1: Loading the Array-Based Dataset - Images as Data

We load from sklearn.datasets, building on Iris and California, but now arrays of images. This introduces high-dimensional data.


```python
# Import libraries
import numpy as np
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt

# Load the dataset
digits = load_digits()

# Data and targets as arrays
data = digits.data
targets = digits.target

# Print shapes
print('Data shape:', data.shape)
print('Targets shape:', targets.shape)
```

# Output interpretation

Data: (1797, 64) – 1797 samples, each with 64 features (pixels). Targets: (1797,) – labels 0-9. Notice flat arrays; we'll reshape for images.


### Step 2: Reshaping Arrays - From Flat to Image

Reshape to view as 8x8 images. This shows NumPy flexibility for data transformation, crucial for image ML.


```python
# First sample
first_digit = data[3]
print('Flat shape:', first_digit.shape)

# Reshape to 8x8
image = first_digit.reshape(8, 8)
print('Image shape:', image.shape)
print('Image array:\n', image)
# Display the image
plt.imshow(image, cmap='gray')
plt.title(f'Digit: {data[3]}')
plt.axis('off')
plt.show()
```

# Output interpretation

(64,) to (8,8). Now it's a matrix: rows/columns as height/width. This allows visualization as image.


### Step 3: Visualizing Images - Using imshow

Plot single image. Introduce after reshape because flat arrays can't be shown as images.


```python
# Plot the image
plt.figure(figsize=(4, 4))
plt.imshow(image, cmap='gray')
plt.colorbar()
plt.title(f'Digit: {targets[0]}')
plt.axis('off')
plt.show()
```

# Output interpretation

Grayscale image of '0', colorbar 0-16 intensity. Notice darker pixels (higher values) form the shape. This helps see data as visual patterns.


### Step 4: Flattening Arrays - Back to Features

Flatten for ML input. Shows reverse of reshape, why: models often expect 1D per sample.


```python
# Flatten the image
flat_again = image.flatten()
print('Flattened shape:', flat_again.shape)

# Check equality
print('Same as original?', np.array_equal(flat_again, first_digit))
```

# Output interpretation

(64,), True. Flattening lossless; useful for switching views.


### Step 5: Multi-Dimensional Slicing - Extracting Parts

Slice image parts. Builds on basic slicing, now 2D.


```python
# Top-left 4x4
top_left = image[:4, :4]
print(top_left)

# Plot subset
plt.imshow(top_left, cmap='gray')
plt.title('Top-Left Quarter')
plt.show()
```

# Output interpretation

4x4 array of values. Image shows part of '0'. Useful for focusing on regions, like edges in digits.


### Step 6: Using Loops for Multiple Visualizations - Grid of Images

Loop to show several. Teach loop here for iteration over samples, but note vectorized alternatives where possible.


```python
# Show first 10 digits
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for i, ax in enumerate(axes.flatten()):
    img = data[i].reshape(8, 8)
    ax.imshow(img, cmap='gray')
    ax.set_title(f'Digit: {targets[i]}')
    ax.axis('off')
plt.tight_layout()
plt.show()
```

# Output interpretation

Grid of 10 images: 0 to 9 mix. Notice variations in writing; loops handy for batches, but for large N, consider sampling.


### Step 7: Basic Analysis on Arrays - Pixel Statistics

Aggregate per image or globally. Uses NumPy for insights.


```python
# Mean pixel value per digit class
unique_digits = np.unique(targets)
for digit in unique_digits:
    digit_data = data[targets == digit]
    mean_pixel = np.mean(digit_data)
    print(f'Digit {digit} mean pixel: {mean_pixel:.2f}')
```

# Output interpretation

Varies, e.g., 0 ~4.5, 1 ~3.0. Lower for thinner digits like 1. Gives intuition: 'filled' digits higher intensity.


### Step 8: Functions for Visualization - Reusable Plotter

Function to plot grid. Builds on functions.


```python
def plot_digit_grid(data, targets, n_rows=2, n_cols=5):
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols*2, n_rows*2))
    for i, ax in enumerate(axes.flatten()):
        if i < len(data):
            img = data[i].reshape(8, 8)
            ax.imshow(img, cmap='gray')
            ax.set_title(f'Digit: {targets[i]}')
            ax.axis('off')
    plt.tight_layout()
    plt.show()

# Call with first 10
plot_digit_grid(data[:10], targets[:10])
```

# Output interpretation

Same grid. Function flexible: change rows/cols for more.


## ⚠️ Common Mistakes Box

- Wrong reshape: 8x8 not 64, errors in dims.
- Forgetting cmap='gray': Colors wrong for grayscale.
- Loops without enumerate: Index errors in subplots.
- Slicing 2D wrong: image[0:4,0:4] not [0:4][0:4].
- Aggregates on flat: Mean ok, but visualize reshaped.
- Function no show(): Plot not displayed.
