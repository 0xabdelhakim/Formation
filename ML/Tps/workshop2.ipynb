## Dataset Context
The Titanic dataset contains information about passengers on the RMS Titanic, which sank in 1912 after hitting an iceberg. It includes data on 891 passengers (a sample from the real event).
The target variable is 'Survived' (0 = did not survive, 1 = survived), which we might predict later in machine learning.
The features include:
- PassengerId: Unique ID (integer)
- Pclass: Ticket class (1=upper, 2=middle, 3=lower) (integer)
- Name: Passenger name (string)
- Sex: male or female (string)
- Age: Age in years (float, with some missing)
- SibSp: Number of siblings/spouses aboard (integer)
- Parch: Number of parents/children aboard (integer)
- Ticket: Ticket number (string)
- Fare: Passenger fare (float)
- Cabin: Cabin number (string, many missing)
- Embarked: Port of embarkation (C=Cherbourg, Q=Queenstown, S=Southampton) (string)
This data represents a real historical tragedy and is used to explore patterns, like who was more likely to survive based on class, age, or gender. Think of it as detective work: inspecting the data to understand the story behind the numbers.

### Step 1: Loading the Dataset from a CSV File - Bringing in External Data
We load data from a CSV because real-world datasets often come in files, not built-in like Iris. This builds on last week's loading but uses Pandas directly, which is more common.

```python
# Import pandas if not already (from last week)
import pandas as pd

# Load the Titanic dataset from a URL (or local file if you have it)
# Note: In practice, download from Kaggle or Seaborn, but here we use a direct link for simplicity
titanic_url = 'https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv'
titanic_df = pd.read_csv(titanic_url)

# Display the first 5 rows to confirm loading
titanic_df.head()
```

# Output interpretation
You should see a table with columns like PassengerId, Survived, Pclass, etc., and the first few rows of data. Notice names, ages, and some NaN (missing) in Cabin. This quick look confirms the data loaded correctly.

### Step 2: Inspecting the Structure - Getting the Big Picture
We use .info() and .shape to see the overall structure: how many rows/columns, data types, and non-null counts. This is crucial before any analysis to know what you're working with and spot issues early.

```python
# Get a summary of the DataFrame
titanic_df.info()

# Get the number of rows and columns
print(titanic_df.shape)
```

# Output interpretation
.info() shows <class 'pandas.core.frame.DataFrame'>, 891 entries, 12 columns, with types like int64, float64, object. Non-null counts reveal missing data (e.g., Age has 714 non-null). .shape gives (891, 12), meaning 891 passengers, 12 features. Notice how this highlights potential problems like missing Ages.

### Step 3: Checking for Missing Values - Identifying Gaps in Data
Missing values can skew analysis, so we check .isnull().sum() to count them per column. This step comes after structure inspection because we need to know columns first.

```python
# Count missing values in each column
missing_values = titanic_df.isnull().sum()

# Print the result
print(missing_values)
```

# Output interpretation
You see a series like Age: 177, Cabin: 687, Embarked: 2, others 0. This tells us Cabin is mostly missing (maybe not useful), Age has some gaps (we'll handle later), and Embarked only a few.

### Step 4: Understanding Data Types - Numbers vs. Categories
Data types tell us how to treat columns: numbers for math, objects for categories. We review from .info(), but focus here to interpret why it matters for ML (e.g., categories need encoding later).

```python
# Print data types of each column
print(titanic_df.dtypes)

# Example: Check a specific column
print(titanic_df['Age'].dtype)  # Should be float64
```

# Output interpretation
Series like PassengerId: int64, Name: object, Age: float64. Object means strings or mixed, like Sex: 'male'/'female'. Float for Age allows decimals (e.g., infants). This helps plan: numerical for scaling, categorical for encoding.

### Step 5: Summary Statistics - Quick Numerical Insights
Use .describe() for means, mins, maxes on numbers, and .value_counts() for categories. This gives a feel for distributions without visuals yet.

```python
# Summary for numerical columns
print(titanic_df.describe())

# Value counts for a categorical column
print(titanic_df['Sex'].value_counts())
```

# Output interpretation
.describe() shows count, mean (e.g., average Age ~29.7), std (spread), min/max (Fare up to 512!). For Sex: male 577, female 314. Notice more males, survival mean ~0.38 (38% survived). This hints at imbalances.

### Step 6: Basic Visualizations - Seeing Distributions
Plots help intuit data better than numbers. We introduce Matplotlib/Seaborn here because inspection often needs visuals to spot patterns, like age distribution or survival by class.

```python
# Import plotting libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Set style for better visuals
sns.set(style='whitegrid')

# Histogram for Age
plt.figure(figsize=(8, 5))
sns.histplot(titanic_df['Age'].dropna(), bins=20, kde=False)
plt.title('Distribution of Passenger Ages')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()

# Bar plot for Survival by Sex
plt.figure(figsize=(8, 5))
sns.countplot(x='Sex', hue='Survived', data=titanic_df)
plt.title('Survival by Sex')
plt.show()
```

# Output interpretation
Age histogram: Peaks around 20-30 years, fewer children/elderly. Use dropna() to ignore missing. Survival bar: Females have higher survival bars for 1 (survived) than males. This visually shows 'women first' pattern.

## ⚠️ Common Mistakes Box
- Ignoring missing values: Running .describe() without checking nulls can mislead (e.g., mean Age ignores 177 missings).
- Wrong data types: Treating 'Pclass' as number but it's categorical (1,2,3 are classes, not quantities).
- Plotting without labels: Plots become confusing without title/xlabel.
- Forgetting to import: sns.histplot() fails without import seaborn as sns.
- Running .value_counts() on numbers: Better for categories; for nums use histograms.
- Not handling NaN in plots: histplot on Age without dropna() might error or skew.
